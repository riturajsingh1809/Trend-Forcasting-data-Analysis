```python
import pandas as pd
import numpy as np

# ============================================================
# STEP 1: LOAD CLEANED DATASET
# ============================================================

file_path = "/content/Cleaned_Sample_Dataset.xlsx"

df = pd.read_excel(file_path)

print("Dataset Loaded Successfully!")
print("Shape:", df.shape)
display(df.head())


# ============================================================
# STEP 2: CHECK DATA INFORMATION
# ============================================================

print("\nDataset Information:")
df.info()


# ============================================================
# STEP 3: CHECK COLUMN NAMES
# ============================================================

print("\nColumn Names:")
print(df.columns.tolist())


# ============================================================
# STEP 4: CHECK MISSING VALUES
# ============================================================

print("\nMissing Values:")
print(df.isnull().sum())


# ============================================================
# STEP 5: CHECK DUPLICATE VALUES
# ============================================================

print("\nDuplicate Rows:")
print(df.duplicated().sum())


# ============================================================
# STEP 6: BASIC STATISTICAL ANALYSIS
# ============================================================

print("\nStatistical Summary:")
display(df.describe())


# ============================================================
# STEP 7: CONVERT DATE COLUMN
# ============================================================

df["date"] = pd.to_datetime(
    df["date"],
    errors="coerce"
)

print("\nDate Range:")
print("Minimum Date:", df["date"].min())
print("Maximum Date:", df["date"].max())


# ============================================================
# STEP 8: CHECK UNIQUE VALUES
# ============================================================

print("\nUnique Regions:")
print(df["region"].unique())

print("\nUnique Categories:")
print(df["category"].unique())


# ============================================================
# STEP 9: CREATE YEAR COLUMN
# ============================================================

df["year"] = df["date"].dt.year

print("\nYear-wise Record Count:")
print(df["year"].value_counts().sort_index())


# ============================================================
# STEP 10: CREATE MONTH COLUMN
# ============================================================

df["month"] = df["date"].dt.month

print("\nMonth-wise Record Count:")
print(df["month"].value_counts().sort_index())


# ============================================================
# STEP 11: CREATE YEAR-MONTH COLUMN
# ============================================================

df["year_month"] = df["date"].dt.to_period("M")

print("\nYear-Month Created Successfully")


# ============================================================
# STEP 12: MONTHLY DEMAND ANALYSIS
# ============================================================

monthly_demand = (
    df.groupby("year_month")["demand"]
    .sum()
    .reset_index()
)

monthly_demand["year_month"] = (
    monthly_demand["year_month"].dt.to_timestamp()
)

print("\nMonthly Demand:")
display(monthly_demand.head(12))


# ============================================================
# STEP 13: MONTHLY REVENUE ANALYSIS
# ============================================================

monthly_revenue = (
    df.groupby("year_month")["revenue"]
    .sum()
    .reset_index()
)

monthly_revenue["year_month"] = (
    monthly_revenue["year_month"].dt.to_timestamp()
)

print("\nMonthly Revenue:")
display(monthly_revenue.head(12))


# ============================================================
# STEP 14: YEAR-WISE DEMAND
# ============================================================

yearly_demand = (
    df.groupby("year")["demand"]
    .sum()
    .reset_index()
)

print("\nYear-wise Demand:")
display(yearly_demand)


# ============================================================
# STEP 15: YEAR-WISE REVENUE
# ============================================================

yearly_revenue = (
    df.groupby("year")["revenue"]
    .sum()
    .reset_index()
)

print("\nYear-wise Revenue:")
display(yearly_revenue)


# ============================================================
# STEP 16: REGION-WISE DEMAND
# ============================================================

region_demand = (
    df.groupby("region")["demand"]
    .sum()
    .sort_values(ascending=False)
    .reset_index()
)

print("\nRegion-wise Demand:")
display(region_demand)


# ============================================================
# STEP 17: CATEGORY-WISE DEMAND
# ============================================================

category_demand = (
    df.groupby("category")["demand"]
    .sum()
    .sort_values(ascending=False)
    .reset_index()
)

print("\nCategory-wise Demand:")
display(category_demand)


# ============================================================
# STEP 18: AVERAGE MONTHLY DEMAND
# ============================================================

average_monthly_demand = monthly_demand["demand"].mean()

print("\nAverage Monthly Demand:")
print(round(average_monthly_demand, 2))


# ============================================================
# STEP 19: MAXIMUM MONTHLY DEMAND
# ============================================================

max_demand = monthly_demand["demand"].max()

max_demand_month = monthly_demand.loc[
    monthly_demand["demand"].idxmax(),
    "year_month"
]

print("\nMaximum Monthly Demand:")
print(round(max_demand, 2))
print("Month:", max_demand_month)


# ============================================================
# STEP 20: MINIMUM MONTHLY DEMAND
# ============================================================

min_demand = monthly_demand["demand"].min()

min_demand_month = monthly_demand.loc[
    monthly_demand["demand"].idxmin(),
    "year_month"
]

print("\nMinimum Monthly Demand:")
print(round(min_demand, 2))
print("Month:", min_demand_month)


# ============================================================
# STEP 21: MONTHLY SEASONALITY ANALYSIS
# ============================================================

seasonality = (
    df.groupby("month")["demand"]
    .mean()
    .reset_index()
)

print("\nAverage Demand by Month:")
display(seasonality)


# ============================================================
# STEP 22: PREPARE TIME SERIES DATA
# ============================================================

time_series = monthly_demand.copy()

time_series = time_series.sort_values(
    "year_month"
).reset_index(drop=True)

time_series = time_series.rename(
    columns={"demand": "actual_demand"}
)

print("\nTime Series Data:")
display(time_series.head())


# ============================================================
# STEP 23: TRAIN-VALIDATION SPLIT
# ============================================================

validation_period = 12

train = time_series.iloc[:-validation_period].copy()

validation = time_series.iloc[-validation_period:].copy()

print("\nTraining Data:")
print(train.shape)

print("\nValidation Data:")
print(validation.shape)

print("\nTraining Period:")
print(
    train["year_month"].min(),
    "to",
    train["year_month"].max()
)

print("\nValidation Period:")
print(
    validation["year_month"].min(),
    "to",
    validation["year_month"].max()
)


# ============================================================
# STEP 24: NAIVE BASELINE FORECAST
# ============================================================

last_value = train["actual_demand"].iloc[-1]

validation["naive_forecast"] = last_value

print("\nNaive Forecast Value:")
print(round(last_value, 2))


# ============================================================
# STEP 25: CALCULATE NAIVE MAE
# ============================================================

naive_mae = np.mean(
    np.abs(
        validation["actual_demand"]
        - validation["naive_forecast"]
    )
)

print("\nNaive Baseline MAE:")
print(round(naive_mae, 2))


# ============================================================
# STEP 26: CALCULATE NAIVE MAPE
# ============================================================

naive_mape = np.mean(
    np.abs(
        (
            validation["actual_demand"]
            - validation["naive_forecast"]
        )
        / validation["actual_demand"]
    )
) * 100

print("\nNaive Baseline MAPE:")
print(round(naive_mape, 2), "%")


# ============================================================
# STEP 27: FORECASTING MODEL
# ============================================================

from statsmodels.tsa.holtwinters import ExponentialSmoothing

train_series = train.set_index(
    "year_month"
)["actual_demand"]

model = ExponentialSmoothing(
    train_series,
    trend="add",
    seasonal="add",
    seasonal_periods=12
)

fitted_model = model.fit(
    optimized=True
)

forecast = fitted_model.forecast(
    validation_period
)

validation["model_forecast"] = forecast.values

print("\nModel Forecast:")
display(
    validation[
        [
            "year_month",
            "actual_demand",
            "model_forecast"
        ]
    ]
)


# ============================================================
# STEP 28: MODEL MAE
# ============================================================

model_mae = np.mean(
    np.abs(
        validation["actual_demand"]
        - validation["model_forecast"]
    )
)

print("\nForecast Model MAE:")
print(round(model_mae, 2))


# ============================================================
# STEP 29: MODEL MAPE
# ============================================================

model_mape = np.mean(
    np.abs(
        (
            validation["actual_demand"]
            - validation["model_forecast"]
        )
        / validation["actual_demand"]
    )
) * 100

print("\nForecast Model MAPE:")
print(round(model_mape, 2), "%")


# ============================================================
# STEP 30: MODEL COMPARISON
# ============================================================

comparison = pd.DataFrame({
    "Model": [
        "Naive Baseline",
        "Exponential Smoothing"
    ],
    "MAE": [
        naive_mae,
        model_mae
    ],
    "MAPE": [
        naive_mape,
        model_mape
    ]
})

print("\nModel Comparison:")
display(comparison)


# ============================================================
# STEP 31: SELECT BEST MODEL
# ============================================================

if model_mae < naive_mae:
    best_model = "Exponential Smoothing"
else:
    best_model = "Naive Baseline"

print("\nBest Model:")
print(best_model)


# ============================================================
# STEP 32: REFIT MODEL ON COMPLETE DATA
# ============================================================

complete_series = time_series.set_index(
    "year_month"
)["actual_demand"]

final_model = ExponentialSmoothing(
    complete_series,
    trend="add",
    seasonal="add",
    seasonal_periods=12
).fit(
    optimized=True
)


# ============================================================
# STEP 33: FUTURE FORECAST
# ============================================================

forecast_period = 12

future_forecast = final_model.forecast(
    forecast_period
)

forecast_result = pd.DataFrame({
    "date": future_forecast.index,
    "forecast_demand": future_forecast.values
})

print("\nFuture 12-Month Forecast:")
display(forecast_result)


# ============================================================
# STEP 34: FORECAST UNCERTAINTY
# ============================================================

residuals = (
    complete_series
    - final_model.fittedvalues
)

residual_std = residuals.std()

forecast_result["lower_bound"] = (
    forecast_result["forecast_demand"]
    - 1.96 * residual_std
)

forecast_result["upper_bound"] = (
    forecast_result["forecast_demand"]
    + 1.96 * residual_std
)

forecast_result["lower_bound"] = (
    forecast_result["lower_bound"].clip(lower=0)
)

print("\nForecast With Uncertainty:")
display(forecast_result)


# ============================================================
# STEP 35: FINAL FORECAST SUMMARY
# ============================================================

forecast_summary = pd.DataFrame({
    "Metric": [
        "Best Model",
        "Validation MAE",
        "Validation MAPE",
        "Forecast Horizon",
        "Residual Standard Deviation",
        "Average Monthly Demand",
        "Maximum Monthly Demand",
        "Minimum Monthly Demand"
    ],
    "Value": [
        best_model,
        round(model_mae, 2),
        round(model_mape, 2),
        forecast_period,
        round(residual_std, 2),
        round(average_monthly_demand, 2),
        round(max_demand, 2),
        round(min_demand, 2)
    ]
})

print("\nFinal Forecast Summary:")
display(forecast_summary)


# ============================================================
# STEP 36: CREATE PYTHON ANALYSIS RESULT FILE
# ============================================================

output_file = "/content/Python_Analysis_Result.xlsx"

with pd.ExcelWriter(
    output_file,
    engine="openpyxl"
) as writer:

    df.to_excel(
        writer,
        sheet_name="Cleaned Data",
        index=False
    )

    monthly_demand.to_excel(
        writer,
        sheet_name="Monthly Demand",
        index=False
    )

    monthly_revenue.to_excel(
        writer,
        sheet_name="Monthly Revenue",
        index=False
    )

    yearly_demand.to_excel(
        writer,
        sheet_name="Yearly Demand",
        index=False
    )

    yearly_revenue.to_excel(
        writer,
        sheet_name="Yearly Revenue",
        index=False
    )

    region_demand.to_excel(
        writer,
        sheet_name="Region Analysis",
        index=False
    )

    category_demand.to_excel(
        writer,
        sheet_name="Category Analysis",
        index=False
    )

    seasonality.to_excel(
        writer,
        sheet_name="Seasonality",
        index=False
    )

    train.to_excel(
        writer,
        sheet_name="Training Data",
        index=False
    )

    validation.to_excel(
        writer,
        sheet_name="Validation Data",
        index=False
    )

    comparison.to_excel(
        writer,
        sheet_name="Model Comparison",
        index=False
    )

    forecast_result.to_excel(
        writer,
        sheet_name="Forecast Result",
        index=False
    )

    forecast_summary.to_excel(
        writer,
        sheet_name="Forecast Summary",
        index=False
    )


# ============================================================
# FINAL OUTPUT
# ============================================================

print("\n" + "="*60)
print("PYTHON ANALYSIS COMPLETED SUCCESSFULLY!")
print("="*60)

print("\nOutput File:")
print("Python_Analysis_Result.xlsx")

print("\nBest Model:", best_model)
print("Validation MAE:", round(model_mae, 2))
print("Validation MAPE:", round(model_mape, 2), "%")
print("Forecast Horizon:", forecast_period, "Months")

print("\nAll analysis results are saved in ONE Excel file.")
print("="*60)
```
